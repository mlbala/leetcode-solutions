```python
A = [1,2,3,4]
print(A)
```

```python
# Append - Insert element at end of array - on average : O(1)
A.append(5)

print(A)
```

```python
# Pop - Remove element from end of array - on average : O(1)
A.pop()

print(A)
```

```python
# Insert - Insert element at a specific index(Not at end of array) - on average : O(n)
A.insert(2,6)

print(A)
```

```python
# Modify an element - O(1)
A[0] = 7

print(A)
```

```python
# Accessing an element - O(1)
print(A[2])
```

```python
#Checking if array has an element - O(n)

if 7 in A:
    print(True)
```
